```python
from diffusers import StableDiffusion3Pipeline
import torch
```

```python
pipeline = StableDiffusion3Pipeline.from_pretrained("stabilityai/stable-diffusion-3.5-medium", torch_dtype=torch.float16).to('cuda')
pipeline.load_lora_weights("bhomik7/sd_3_5m_dreambooth_lora_ft_FULL_RES", weight_name='pytorch_lora_weights.safetensors')
```

```python
prompt ="""A vibrant and energetic music app UI inspired by music festivals. The design incorporates bright colors, playful icons, and dynamic animations. The playback screen features a glowing soundwave visualizer in the background, along with festival-themed illustrations like stages, lights, and crowds. The playlist view is colorful and engaging, with icons representing different genres and moods."""
```

```python
image = pipeline(prompt = prompt, 
                 guidance_scale = 15, 
                 num_inference_steps = 50,
                 height = 1024,
                 width = 1024,
                negative_prompt="bad text, gibberish text, distorted image, distorted figure, distorted text",
                generator=torch.manual_seed(2957138076)).images[0]
```

```python
image
```
