```python
from diffusers import StableDiffusion3Pipeline
import torch
```

```python
pipeline = StableDiffusion3Pipeline.from_pretrained("stabilityai/stable-diffusion-3.5-medium", torch_dtype=torch.float16).to('cuda')
# pipeline.load_lora_weights('bhomik7/sd_3_5m_dreambooth_lora_ft2', weight_name='pytorch_lora_weights.safetensors')
```

```python
prompt ="""Create a modern and immersive music app interface that feels sleek and engaging. The home screen should feature personalized recommendations, recently played tracks, and trending albums, all arranged in an intuitive, scrollable layout. A search bar should be placed at the top for quick music discovery. The 'Now Playing' screen should focus on a large album art display at the center, surrounded by clean controls for play, pause, next, and previous tracks, as well as shuffle and repeat options. Below the album art, include a progress slider with time indicators showing elapsed and remaining playback. The playlist view should offer a simple yet elegant list of tracks with titles, durations, and a 'Play All' button, allowing users to navigate effortlessly. Use a dark theme with vibrant neon accents, like blue or purple, and smooth gradients to create a visually captivating experience. Icons and typography should be minimalist and cohesive, ensuring the interface feels modern and intuitive, optimized for an enjoyable music-listening journey."""
```

```python
image = pipeline(prompt = prompt, 
                 guidance_scale = 15, 
                 num_inference_steps = 50,
                 height = 1024,
                 width = 1024,
                negative_prompt="bad text, gibberish text, distorted image, distorted figure, distorted text",
                generator=torch.manual_seed(2957138076)).images[0]
```
